```python
"""Use Case 2: Salary Processing System (HR / Payroll Automation)
Demonstrates input/output functions, default arguments, arbitrary args, returning values, and composing multiple functions.
What Problem It Solves:
Companies calculate employee salary differently depending on bonus percentage, incentives, PF, tax, etc.
How Function-Based Programming Helps:
We create multiple small reusable functions (bonus calc, tax calc, net salary calc)
Then compose them together.
"""
salary=15000
hikepercent=23
taxpercent=12

def bonus_calc(salary,hikepercent):
    return salary*(hikepercent/100)

def tax_calc(salary,taxpercent):
    return salary*(taxpercent/100)

def net_salary(salary):
    bonus=bonus_calc(salary,hikepercent)
    tax=tax_calc(salary,taxpercent)
    return salary+bonus-tax

net_salary(salary)
```

```python
"""Usecase 2: IT Gross Sal Calculator - **kwargs
You are working as a software developer for an HR analytics firm (like Aon Hewitt / HRworkways). Different IT companies structure their annual employee payouts differently based on available compensation components:

CTS pays Base Salary + Bonus Percentage + Fixed Incentives.
INFY pays Base Salary + Bonus Percentage (No incentives).
HCL pays Base Salary only (No bonus, no incentives).
Because each company provides a different combination of compensation parameters, write a single function using **kwargs (arbitrary keyword arguments) to dynamically compute the total gross salary based on whichever components are passed in."""



def salary(**kwargs):
    for i,j in kwargs.items():
        if i=="cts":
            salary=j[0]
            hikepercent=j[1]
            incentive=j[2]
            bonus=salary*(hikepercent/100)
            print("\nSalary",salary+bonus+incentive)
        elif i=="infosys":
            salary=j[0]
            hikepercent=j[1]
            incentive=0
            bonus=salary*(hikepercent/100)
            print("\nSalary",salary+bonus+incentive)
        elif i=="hcl":
            salary=j
            hikepercent=0
            incentive=0
            bonus=salary*(hikepercent/100)
            print("\nSalary",salary+bonus+incentive)
        else:
            print("Done")

salary(cts=[10000,10,4000],infosys=[20000,10],hcl=30000)
```

```python
#Usecase 2: IT Gross Sal Calculator - **kwargs
#CTS -> sal, bon, inc (100000,5000,.02)
#INFY -> sal, bon (100000,5000)
#Can i achieve using arb arg fun? yes, or which one is better? KWArgs is better because we have to handle arguments using name, not by using position.
def hr_workways_sal_fun(**kwargs):
    print(kwargs)
    compname=kwargs['compname']
    if compname in ['CTS','INFY','HCL']:
        sal=kwargs['sal']
        bon=kwargs.get('bon',0)
        inc=kwargs.get('inc',0)
        inc_amt=sal*inc
        return int(sal+bon+inc_amt)

#Emp1
final_sal_cts=hr_workways_sal_fun(compname='CTS',sal=100000,bon=5000,inc=.02)
final_sal_infy=hr_workways_sal_fun(compname='INFY',sal=100000,bon=5000)
final_sal_hcl=hr_workways_sal_fun(compname='HCL',sal=100000,inc=.03)
final_sal_wipro=hr_workways_sal_fun(compname='WIPRO',sal=100000,bon=6000)

print("cts final sal",final_sal_cts)
print("infy final sal",final_sal_infy)
print("hcl final sal",final_sal_hcl)
print("wipro final sal",final_sal_wipro)
```

```python
"""Use Case 2: Safe Integer Input
Ask the user to enter a number.
Try converting it to an integer.
If conversion fails, print:
"Invalid input. Please enter a numeric value."
"""

value=input("Enter a number : ")
try:
    value=int(value)#typecasting
    print(value)
except ValueError as e:
    print("Invalid input. Please enter a numeric value.")
```

```python
"""Use Case 3 (Bug Fixing): Multiple Exception Handling
 Fix the below code so it handles both invalid input and division by zero correctly.
Incorrect code:
num1 = int(input("Enter number 1: "))
 num2 = int(input("Enter number 2: "))
 result = num1 / num2
 print("Result:", result)
Expected behavior:
If user enters non-numeric values → print "Invalid input"
If num2 is zero → print "Cannot divide by zero."
Otherwise print the result.
"""
try:
    num1 = int(input("Enter number 1: "))
    num2 = int(input("Enter number 2: "))
    result = num1 / num2
    print("Result:", result)
except ValueError as e:
    print("Invalid input. Please enter a numeric value.")
except ZeroDivisionError as e:
    print("Division by zero is not allowed, Try entering the 2nd number/value greater than 0")
```
